```python
# Exercício 6 - Crie um programa que leia um número e informe se ele é par ou ímpar.
def imprime(verifica):
    try:
        if(verifica == 0):
            print('Número é par')
        else:
            print('Numero é impar')
    except ValueError:
            print('Informe um número válido')
           

try:
     
    numero = input('Informe um número')
    num = int(numero)
    verifica = num % 2
    imprime(verifica)
 
except ValueError:
        print('Informe um número válido')
```

```python
#potência

def potencia(base, expoente):
    return base ** expoente


numero = int(input('Informe o primeiro número de base para cálcularmos a potência'))
numero2 = int(input('Informe o expoente para cálcularmos a potência'))

print(potencia(numero,numero2))
```

```python

# Exercício 7 - Crie uma função que receba dois números e retorne o maior deles.
def maior(n1, n2):
    if(n1 > n2):
        print(f"O primeiro número que é  {n1} é maior que {n2}")
    if(n1 < n2):
        print(f"O segundo número que é {n2} é maior que {n1}")
    if(n1 == n2):
        print(f"Numeros iguais  {n1}")
    

try:
    valor1 = int(input('Informe o primeiro númeor'))
    valor2 = int(input('Informe o segundo número'))
    maior(valor1,valor2)
except ValueError:
    print('Informe números válidos')
```
