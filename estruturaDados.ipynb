```python
#listas colchetes
frutasL = ["maçã", "manga", "acerola", "amora"]

#tupla parenteses
frutasT =()

#dicionario abre e fecha chaves
frutasD = []

print(frutasL)
print(frutasT)
print(frutasD)
```

```python
#adicionar item na lista
frutasL.append('banana')
print(frutasL)

frutasL.append('uva')
print(frutasL)

#ACESSANDO ITEM DA LISTA - é PELO INDICE
print(frutasL[0])

#ultimo elemento da lista
print(frutasL[-1])

#removendo item da lista
frutasL.remove('banana')
print(frutasL)

#alterar a posição zero que é maça para morango
frutasL[0]='morango'

print(frutasL)
```

```python
#Tuplas
dias_semana =("domingo", "segunda", "terça", "quarta", "quinta", "sexta", "sábado")
print(dias_semana)
```

```python
#Dicionário de dados
aluno={"nome": "BOB", "idade":22, "curso":"Datascience academy", "aluno_ativo":True}

print(aluno)
print(aluno["nome"] )
print(aluno.get("curso"))

#adiciona uma chave ao dict
aluno['cidade'] = 'São Paulo'

#modificar a chave
aluno['nome']='MARIA'

#remover uma chave
del(aluno['aluno_ativo'])
print(aluno)


```

```python
#Conjuntos (Sets)
#com chaves. DIcionário tbm são chaves. Mas dicionário são chave e valor.
numeros = {1,2,3,3,4,5,6,7,8,4}
#ao imprimir, remove os repetidos.
print(numeros)
```

```python
#Conversão de tipos
valor = '123'
conversao1 = int(valor)
print(conversao1)
```

```python
valor='123'
conversao2 = float(valor)
print(conversao2)
```

```python
#convertendo estrutura de dados

#lista com valores duplicados
numeros = [1,2,3,4,4,5,6,7,9]

#converter em conjunto - Set(numeros) - remove duplicatas
numeros2 = set(numeros)

#convertendo o conjunto de volta para lista
numeros3 = list(numeros2)
print(numeros3)

#útil quando queremos fazer limpeza de dados.
```
